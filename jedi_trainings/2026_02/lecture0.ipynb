# Introduction

Three lectures are available under the j`ed_trainings/2026-02` directory. These lectures were developed for the February 2026 training held at GSFC. The notebooks are intended to be used as static materials, and the input files required to run the examples in each lecture are not included in this GitHub repository to keep the repository size manageable.

Copies of the required input files are available on DISCOVER at `/discover/nobackup/mabdiosk/JEDI_practicals/live_demo`, which you can copy and use.

The practical examples available in this repository are designed for users to work through independently. They include low-resolution input files along with instructions on how to run them.

We hold regular office hours to answer questions related to the tutorials and JEDI. Please contact Maryam Abdi-Oskouei if you would like to attend the office hours.

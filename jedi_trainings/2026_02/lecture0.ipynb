# Introduction

Welcome to the February 2026 JEDI training!

This training contains three lectures developed for the JEDI training held at GSFC in Feburary 2026. The notebooks [JEDI trainings (2026-02)](https://mer-a-o.github.io/howtojedi) are designed as guided reading materials — you don't need to run them to follow along, and to keep the repository lightweight, the input files for the live examples aren't included here. If you'd like to explore those, copies of everything you need are available on DISCOVER at `/discover/nobackup/mabdiosk/JEDI_practicals/live_demo`.

For hands-on practice, the [JEDI hands-on examples](https://mer-a-o.github.io/howtojedi/) is a great place to start. These examples come with low-resolution input files and step-by-step instructions so you can work through them at your own pace.

And don't hesitate to reach out. We hold regular office hours to answer questions about the tutorials and JEDI in general.
